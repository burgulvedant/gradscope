```python
numeric_cols = df.select_dtypes(include="number").columns

outlier_summary = {}

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outlier_summary[col] = ((df[col] < lower) | (df[col] > upper)).sum()

pd.Series(outlier_summary).sort_values(ascending=False)
```

```python
flagged_cols = pd.Series(outlier_summary)
flagged_cols = flagged_cols[flagged_cols > 0].index

df[["city"] + list(flagged_cols)]
```

```python
for col in flagged_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower) | (df[col] > upper)][["city", col]]
    
    print(f"\n{col}")
    print(outliers.to_string(index=False))
```

```python
df["numbeo_update_date"] = pd.to_datetime(
    df["numbeo_update_date"],
    format="%d-%b-%Y"
)

df[["city", "numbeo_update_date"]]
```
